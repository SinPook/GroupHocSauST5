```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# Các thông số của bài toán
mu = 1.0
sigma = 1.0
N = 200  # Số lượng điểm ngẫu nhiên

x_min = mu - 5 * sigma  # -4.0
x_max = mu + 5 * sigma  #  6.0

# Sinh ngẫu nhiên N điểm x trong đoạn [x_min, x_max]
x = np.random.rand(N) * (x_max - x_min) + x_min

# Tính giá trị y theo công thức Phân phối chuẩn N(mu, sigma^2)
y = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-((x - mu) ** 2) / (2 * sigma**2))

print("b. Sinh cặp (x, y) với y làm hàm phân phối chuẩn")
print(f"+ x: {x_min} <= min [{np.min(x): .2f}] <= max [{np.max(x): .2f}] <= {x_max}")
print(f"+ Ve ham so y(x)")
plt.figure(figsize=(6,6))      # kich thuoc ban ve
plt.scatter(x, y)              # ve cac diem (xi, yi)
plt.savefig("4b.png")          #  luu do thi
plt.show()                     #  hien thi do thi
```

```python
""" CÁC BẠN LÀM BÀI Ở ĐÂY """
# Sinh dãy 100.000 phần tử theo phân phối chuẩn N(mu=1.0, sigma=1.0)
x = np.random.normal(loc=1.0, scale=1.0, size=100000)

# Gom nhóm dữ liệu thành các khoảng nhỏ và đếm số lượng phần tử trong mỗi khoảng
cnt, val_edges = np.histogram(x, bins=100)
# Tính giá trị trung tâm cho từng khoảng để làm trục X (val)
val = (val_edges[:-1] + val_edges[1:]) / 2

print( "c. Sinh day so theo phan phoi chuan")
print(f"+ x: len = {len(x)}, min [{np.min(x)}], max [{np.max(x)}]")
print(f"+ Mat so xac suat cua day so x")
plt.figure(figsize=(6,6))      # kich thuoc ban ve
plt.xlim(np.min(val), np.max(val))
plt.scatter(val, cnt)
plt.show()
```

### Bài 3. Tính gần đúng số pi dùng phương pháp Monte Carlo

Vẽ một đường tròn $C$ có bán kính $r$ nội tiếp một hình vuông $C$, và đặt ngẫu nhiên $N$ chấm lên hình vuông. Tỉ lệ các chấm nằm trong hình tròn trên tổng số chấm xấp xỉ bằng diện tích của hình tròn chia cho hình vuông như sau:
$$
\frac{N_C}{N} \approx  \frac{\mathbb{S}_C}{\mathbb{S}_N}
$$
nơi mà $N_C$ là các chấm trong hình tròn, $N$ là tổng số chấm, $\mathbb{S}_C$ và $\mathbb{S}_N$ lần lượt là diện tích của hình tròn và hình vuông.
$$ \mathbb{S}_C = \pi * r^2 $$
$$ \mathbb{S}_N = (2*r)^2 = 4r^2 $$
Do đó, công thức trên trở thành:
$$
\frac{N_C}{N} \approx \frac{\pi * r^2}{4r^2} = \frac{\pi}{4}
$$
Suy ra, ta có giá trị số $\pi$ được tính như sau:
$$
\pi \approx \frac{4N_C}{N}
$$

Các bạn hãy lập trình tính sấp xỉ số $\pi$ dùng phương pháp Monte Carlo. Cho biết các sai số với $N = 100$, $N = 10000$, và $N = 1000000$. Vẽ đồ thị minh họa cho chương trình.
