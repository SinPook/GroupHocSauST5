```python
# ex2_scientific1

# Nội dung bài
# Hiển thị ngẫu nhiên N=50 hình tròn:
# + Tâm $c_i = (x_i, y_i)$ ngẫu nhiên với $x_i, y_i \in [0, 1]$
# + Màu sắc $color_i$ ngẫu nhiên với $color_i \in [0, 1]$
# + Diện tích $area_i = \pi * r_i^2$ với $r_i$ ngẫu nhiên
```

```python
# Import thư viện
import numpy as np
import matplotlib.pyplot as plt

# Sinh ngẫu nhiên N = 50 hình tròn
N = 50
x = np.random.rand(N)        # tọa độ tâm x_i trong [0, 1]
y = np.random.rand(N)        # tọa độ tâm y_i trong [0, 1]
colors = np.random.rand(N)   # màu sắc color_i trong [0, 1]
r = np.random.rand(N)        # bán kính r_i trong [0, 1]
area = np.pi * r**2         # diện tích area_i = pi * r_i^2

# Kiểm tra dữ liệu
assert len(x) == N and len(y) == N
assert len(colors) == N and len(r) == N and len(area) == N
assert np.all((x >= 0) & (x <= 1))
assert np.all((y >= 0) & (y <= 1))
assert np.all((colors >= 0) & (colors <= 1))
assert np.all((r >= 0) & (r <= 1))

print(f'+ x     : len={len(x)}, khoảng [{x.min():.3f}, {x.max():.3f}]')
print(f'+ y     : len={len(y)}, khoảng [{y.min():.3f}, {y.max():.3f}]')
print(f'+ color : len={len(colors)}, khoảng [{colors.min():.3f}, {colors.max():.3f}]')
print(f'+ r     : len={len(r)}, khoảng [{r.min():.3f}, {r.max():.3f}]')
print(f'+ area  : len={len(area)}, khoảng [{area.min():.3f}, {area.max():.3f}]')

# Hiển thị 50 hình tròn
fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(x, y, s=area * 5000, c=colors, cmap='viridis', alpha=0.6)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect('equal', adjustable='box')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('N = 50 hình tròn ngẫu nhiên')
ax.grid(True, alpha=0.25)
plt.show()
```
